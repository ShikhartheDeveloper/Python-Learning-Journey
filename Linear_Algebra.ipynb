** Norm 2 Euclidean lenght**
firstly do the square of all values in vector then add  them and then square it

||v2|| = √(v1**2 + v2**2 + ..... + Vn**2)

```python
z = np.array([1,2,2])
print(np.linalg.norm(z))

# if we find the euclidean distance from 2 points then

h = np.array([2,3])
i = np.array([4,5])
print(np.linalg.norm(i -h))
```

***Norm 1 (MANHATTAN DISTANCE) :-    here is not taking the square and square root too***

*||v|| = |v1| + |v2| + .... + |vn|*

if there is any  negative values so it turns to absolute value like -7 => 7

```python
a = np.array([3,4,-2])    # -2 positive hojayga = 2
print(np.linalg.norm(a, ord=1))
```

***Norm ∞ (Max/ infinity norm) :-    here is not taking the square and square root too but here find the max value (absoulute) from vectors***

||v||∞ = max(|v1|,|v2|,.......,|vn|)

```python
k = np.array([4,-9,7])  # -9 turns absolute becomes 9
print(np.linalg.norm(k, ord=np.inf))
```
