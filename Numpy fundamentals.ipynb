```python
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
```

```python
print(np.sum(a1))
# gives you the max element present in the array   max/min/sum/prod
print(np.max(a1))
print(np.min(a1))
# np.prod() multiplies all elements in the array together and returns a single number (the product).
print(np.prod(a1))
```

```python
# if we want to get  Max of Every Column → axis=0
# if we want to get  Max of Every Row → axis=1
print(a1)
print(np.max(a1,axis=0))
print(np.max(a1,axis=1))
```

```python
# mean/median/std/var
# mean -> Average [1,2,3,4,5] sum = 15 , total number = 5 so mean is 15/5 = 3
np.mean(a1)

# np.median() finds the middle value of a sorted array — it's less affected by extreme outliers than np.mean().
np.median(a1)

# .std() / np.std() — Standard Deviation

# std measures how spread out the values in an array are from the mean. A low std means values are close to the average; 
# a high std means they're spread far apart.
np.std(a1)

# .var() / np.var() — Variance

# var measures how spread out values are from the mean — same idea as std, but without the final square root step.
#     It's the "average squared distance from the mean."

np.var(a1)
```

# Stacking
* These functions combine multiple arrays together — the difference is the direction they stack in.
* np.vstack() — Stack Vertically (adds rows, stacks on top of each other)
* np.hstack() — Stack Horizontally (adds columns, side by side)

```python

a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)
print(a4)
print(a5)
```

```python
# horizontal stacking
np.hstack((a4,a5))
```

```python
# vertical stacking
np.vstack((a4,a5))
```

# Splitting
* Splitting is the opposite of stacking — instead of combining arrays, you break one array into multiple smaller arrays.
* This splits arr into 3 equal parts. Note the result is a list of arrays, not one array.

```python
np.hsplit(a4,2)
```
